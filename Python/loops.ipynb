### Iteration

```python
#Iterate over a list of integers

list_1 = [10,20,30,40,50]

for num in list_1:
    print(num)
```

```python
#Iterate over a String

str_1 = "Odisha"

for ch in str_1:
    print(ch)
```

```python
#Iterate over a dictionary

students_data = {1: ['Sam', 25], 2:['Sharma',28], 3:['Ravi',22], 4:['Sid',23], 5:['Varun',24]}

for i,j in students_data.items():
    print(i,j)

for keys in students_data.keys():
    print(keys)

for value in students_data.values():
    print(value)
```

### Comprehensions

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append(i)

list_2
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append(len(i))

list_2
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = [len(i) for i in list_1]

list_2
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

dict1 = {i: len(i) for i in list_1}

dict1
```
