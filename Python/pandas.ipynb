## Concat & Merging

```python
df_dict = pd.DataFrame({"A":["A0","A1","A2","A3"],
                       "B":["B0","B1","B2","B3"],
                       "C":["C0","C1","C2","C3"],
                       "D":["D0","D1","D2","D3"]}, 
                        index=[0,1,2,3]
                      )
```

```python
df_dict2 = pd.DataFrame({"A":["A4","A5","A6","A7"],
                         "B":["B4","B5","B6","B7"],
                         "C":["C4","C5","C6","C7"],
                         "D":["D4","D5","D6","D7"]},
                        index=[4,5,6,7]
                       )
```

```python
result = pd.concat([df_dict,df_dict2])
```

```python
df_dict
```

```python
df_dict2
```

```python
result
```

```python
df_dict2 = pd.DataFrame({"E":["A4","A5","A6","A7"],
                         "F":["B4","B5","B6","B7"],
                         "C":["C0","C1","C2","C3"],
                         "D":["D0","D1","D2","D3"]},
                       )
```

```python
df_merge = pd.merge(df_dict, df_dict2)
```

```python
df_merge
```

```python
pd.merge(df_dict,df_dict2,on=["C","D"],how="right")
```
