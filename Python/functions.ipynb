```python
def even_odd(x):
    if x % 2 == 0:
        print(x, "is even number")
    else:
        print(x, "is odd number")

even_odd(2)
```

### Factorial

```python
def fact_number(n):
    fact = 1
    for i in range (1, n+1):
        fact = fact * i
    print(fact)

fact_number(5)
```

### Sum of all natural numbers till the number provided

```python
def sum_number(n):
    sum = 0
    for i in range(1,n+1):
        sum = sum + i
    print(sum)

sum_number(3)
```

### Palindrome Problem

```python
def palindrome(str):
    return str == str[::-1]

print(palindrome("madam"))
print(palindrome("ono"))
print(palindrome("exe"))
print(palindrome("time"))
```

### Fibonacci Series

```python
def fibo(n):
    a, b = 0, 1
    for i in range(n):
      print(a, end=" ")
      a,b = b, a + b

fibo(8)
```
