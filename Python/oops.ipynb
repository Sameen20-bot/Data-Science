```python
class Area:
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def area_calculate(self):
        return self.length * self.breadth

area = Area(2,2)
area.area_calculate()
```

```python
class Shape:
    def __init__(self,color):
        self.color = color
    def calculate_area(self):
        pass
    def price(self):
        color_shape = {'red': 10, 'green': 20, 'blue': 30}
        return self.calculate_area() * color_shape[self.color]
        
```

```python
class Circle(Shape):
    PI = 3.14
    def __init__(self,radius):
        self.radius = radius
    def calculate_area(self):
        return Circle.PI * (self.radius ** 2)
```

```python
c = Circle(7)
c.color = 'green'
print(c.price())
```

```python
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def calculate_area(self):
        return self.length * self.breadth
    def __str__(self):
        return "Area of rectangle= " + str(self.calculate_area())
```

```python
rect = Rectangle(10,10)
rect.__str__()
```
