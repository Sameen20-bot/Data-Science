### Reading a file

```python
file = open('test.txt','r')

for line in file:
    print(line)
```

```python
file = open('test.txt','r')

print(file.read())
```

```python
file = open('test.txt','r')
print(file.read(5))
```

### Writing a file

```python
file = open("test.txt","w")

file.write("Bye, byeee")

file.close()
```

```python
file = open("test.txt","r")
print(file.read())
```

```python
with open("test.txt",'w') as file:
    file.write("pee poo poo")
```

```python
with open("test.txt","r") as file:
    print(file.read())
```

```python
with open("test.txt","a+") as file:
    file.write("\n")
    file.write("tyee tyeee")
    file.close()
```

```python
file = open("test.txt","r")
print(file.read())
```
