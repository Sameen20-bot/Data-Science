### Area of a circle

```python
import math

def area(r):
    return math.pi*(r**2)

radii = [1,2,3,4]
areas = []

for r in radii:
    a = area(r)
    areas.append(a)

print(areas)
```

### map(f,iterable object)

```python
list(map(area,radii))
```

### Convert Celcius scale to Farheneit
### F = 9/5*C + 32

```python
temps = [("Karachi", 35), ("Berlin", 12), ("Tokyo", 25), ("Hong Kong", 31), ("Sydney", 30), ("Abu Dhabi", 35)]
```

```python
cel_to_fah = lambda data: (data[0], 9/5 * data[1] + 32)
list(map(cel_to_fah,temps))
```

### Filter Function - filters out the data
### filter(f, data)

```python
import statistics
data = [10,20,30,40,50,60]
```

```python
avg = statistics.mean(data)
print(avg)
```

```python
list(filter(lambda x: x > avg, data))
```

```python
name = ["Sameen", "Zaki", "Daniya", 0, 0.0, "", "Qudsia"]
```

```python
list(filter(None,name))
```

```python
number = [12, 15, 20, 0, 0.0]
```

```python
list(filter(None,number))
```

### Reduce Function
### reduce(f, data)

```python
from functools import reduce
```

```python
data = [1,2,3,4,5]

reduce(lambda x,y: x * y, data)
```
