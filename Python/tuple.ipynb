### Tuples can be defined in various ways

```python
t = 1,2,3
type(t)
```

```python
t2 = 1,2,3,
type(t2)
```

```python
t3 = 1,
t4 = 1.1,
type(t3)
```

```python
type(t4)
```

```python
t5 = (1)
t6 = (1,)

print(type(t5))
print(type(t6))
```

### min, max, sum functions

```python
t = (2,4,6,8)
sum(t)
```

```python
t = (2,4,6,8,'s')
sum(t)
```

```python
t = (2,4,6,8)
print(max(t))
print(min(t))
```

### Immutability

```python
t = ("Hello!!", "Welcome" , 5, 10, "world", "of", "Analytics!!")
t[0] = "bye"
```

```python
print(t[0:2] + ("to","the") + t[-3:])
```

```python
t = ("disco", 12, 4.5)
print(t[0][2])
```

### Sorting a tuple

```python
t = (2,3,8,5,4,1)
b = sorted(t)
a = tuple(b)
print(a)
```
