```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Data = {'Year': [1920,1930,1940,1950,1960,1970,1980,1990,2000, 2010, 2020],
       'Exchange Rate': [65, 69, 71, 64, 62, 59, 72, 71, 75, 78, 81]}
df = pd.DataFrame(Data)
```

```python
df.plot(x="Year",y="Exchange Rate",kind="barh")
plt.show()
```

```python
df.plot(x="Year",y="Exchange Rate",kind="scatter")
plt.show()
```

```python
plt.scatter(df["Year"],df["Exchange Rate"])
```

```python
x = np.array([[1,2,3]])
y = np.array([[4,5,6]])

plt.scatter(x,y)
plt.show()
```

```python
churn = pd.read_csv("churn.csv")
```

```python
churn.Geography.value_counts().plot(kind="barh")
```

```python
plt.bar(churn["Geography"],churn["Age"].max(),color="green",edgecolor="black")
plt.title("Geograph As Age",fontdict={"fontsize":20,"color":"black"})
plt.show()
```

```python
plt.scatter(churn.Geography,churn.Age)
plt.title("Age And Geography Scatter")
plt.xlabel("This is X axis")
plt.ylabel("This is Y axis")
plt.show()
```

```python
plt.hist(churn.Age, bins=30)
plt.ylabel('Probability')
plt.xlabel('Age')
plt.show()
```

```python
churn.Tenure.plot.box()
```

```python
labels = ["Python", "Machine Learning", "Deep Learning", "Scikit"]
explode = (0,0,0.2,0)
sizes = [20,30,30,10]
fig1,ax1 = plt.subplots()
ax1.pie(sizes, explode=explode, labels=labels,shadow='True',startangle=90)
ax1.axis('equal')
plt.show()
```
