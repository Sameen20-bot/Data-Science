### Set Operations

```python
A = {0,2,4,6,8}
B = {1,2,3,4,5}
```

```python
print(A | B)
print(A.union(B))
```

```python
print(A & B)
print(A.intersection(B))
```

```python
print(A - B)
print(A.difference(B))
```

```python
### symmetric
print(A ^ B)
```
