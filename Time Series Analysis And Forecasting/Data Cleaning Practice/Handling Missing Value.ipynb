# Handling Missing Values

## 1. Deleting the columns with missing Data (yaha pa nhi krna kyu kay kam hain null values but still for practice)

```python
updated_df = df.dropna(axis = 1)
```

```python
updated_df.info()
```

## 2. Deleting the rows with missing values (kam null values hain this approach is good)

```python
updated_df1 = df.dropna(axis = 0)
```

```python
updated_df1.info()
```

## 3. Filling the missing values imputent

```python
updated_df3 = df
updated_df3["Age"] = updated_df3["Age"].fillna(df["Age"].mean())
```

```python
updated_df3.info()
```

## 4. Forward and Backward Fill

```python
updated_df3["Gender"] = updated_df3["Age"].bfill()
```

```python
updated_df3.info()
```
