```python
df_Train = pd.read_csv("DailyDelhiClimateTrain.csv")
df_Test = pd.read_csv("DailyDelhiClimateTest.csv")
```

# Uni-Variate Forecasting

```python
df_Train = df_Train.rename(columns = {'meantemp':'y', 'date':'ds'})
df_Train['y_orig'] = df_Train['y']
df_Train['y'] = np.log(df_Train['y'])
```

```python
df_Train.head()
```

```python
model = Prophet()
model.fit(df_Train)
```

```python
future_Data = model.make_future_dataframe(periods = 113, freq = 'D')
forecast_Future = model.predict(future_Data)
```

```python
model.plot(forecast_Future);
```

```python
model.plot_components(forecast_Future);
```

```python
# Univariate: y ko log kiya tha, isliye exp karke wapas real temperature
# .copy() zaroori hai - warna forecast_Future hi modify hota hai aur
# cell dobara chalane par exp DO BAAR lag jata hai
forecast_data_orig = forecast_Future.copy()
forecast_data_orig["yhat"] = np.exp(forecast_data_orig["yhat"])
forecast_data_orig["yhat_lower"] = np.exp(forecast_data_orig["yhat_lower"])
forecast_data_orig["yhat_upper"] = np.exp(forecast_data_orig["yhat_upper"])
```

```python
forecast_data_orig.head()
```

```python
model.plot(forecast_data_orig)
```

```python
df_Train["y_log"] = df_Train["y"]
df_Train["y"] = df_Train["y_orig"]
```

```python
final_df = pd.DataFrame(forecast_data_orig)
```

```python
actual_chart = go.Scatter(y=df_Train["y"], name= 'Actual')
predict_chart = go.Scatter(y=final_df["yhat"], name= 'Predicted')
predict_chart_upper = go.Scatter(y=final_df["yhat_upper"], name= 'Predicted Upper')
predict_chart_lower = go.Scatter(y=final_df["yhat_lower"], name= 'Predicted Lower')
py.plot([actual_chart, predict_chart, predict_chart_upper, predict_chart_lower], image_width=400, image_height=400)
```

# Multi Variat Forecasting

```python
df_Train_Multi = pd.read_csv("DailyDelhiClimateTrain.csv")
df_Test_Multi = pd.read_csv("DailyDelhiClimateTest.csv")
df_Train_Multi.head()
```

```python
df_Test_Multi = df_Test_Multi.rename(columns = {'meantemp':'y', 'date':'ds'})
df_Test_Multi['y_orig'] = df_Test_Multi['y']
```

```python
df_Train_Multi = df_Train_Multi.rename(columns = {'meantemp':'y', 'date':'ds'})
df_Train_Multi['y_orig'] = df_Train_Multi['y']
```

```python
model_new = Prophet()
model_new.add_regressor("humidity")
model_new.add_regressor("wind_speed")
model_new.add_regressor("meanpressure")
model_new.fit(df_Train_Multi)
```

```python
# Regressors wale model mein make_future_dataframe() kaam nahi karta,
# kyunki future ke humidity/wind_speed/meanpressure Prophet khud nahi bana sakta.
# Isliye train + test ko jod kar future dataframe khud banate hain.
future_Data_New = pd.concat([df_Train_Multi, df_Test_Multi], ignore_index=True)
future_Data_New = future_Data_New[['ds', 'humidity', 'wind_speed', 'meanpressure']]
future_Data_New['ds'] = pd.to_datetime(future_Data_New['ds'])

# Sanity check - NaN nahi hona chahiye
print(future_Data_New.shape)
print(future_Data_New.isnull().sum())

forecast_Future_New = model_new.predict(future_Data_New)
forecast_Future_New[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
model_new.plot(forecast_Future_New);
```

```python
model_new.plot_components(forecast_Future_New);
```

```python
# NOTE: Multivariate mein np.log NAHI kiya tha (cell upar dekho),
# isliye yahan np.exp() ki zarurat NAHI hai.
# yhat already real temperature (Celsius) mein hai.
#
# Agar exp lagate to exp(25) = 72,000,000,000 aata - bilkul galat.
forecast_Future_New[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].head()
```

```python
model_new.plot(forecast_Future_New);
```

```python
# Actual vs Predicted - Multivariate
df_All_Multi = pd.concat([df_Train_Multi, df_Test_Multi], ignore_index=True)

actual_chart        = go.Scatter(y=df_All_Multi["y_orig"],        name='Actual')
predict_chart       = go.Scatter(y=forecast_Future_New["yhat"],       name='Predicted')
predict_chart_upper = go.Scatter(y=forecast_Future_New["yhat_upper"], name='Predicted Upper')
predict_chart_lower = go.Scatter(y=forecast_Future_New["yhat_lower"], name='Predicted Lower')
py.plot([actual_chart, predict_chart, predict_chart_upper, predict_chart_lower])
```
