```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.datasets import make_regression

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

from sklearn.ensemble import BaggingRegressor, RandomForestRegressor

from sklearn.tree import DecisionTreeRegressor

from sklearn.svm import SVR
```

```python
### n_informative make_classification() mein batata hai ke total features mein se kitne 
### features actually target (y) ko predict karne ke liye useful/informative hain.
```

```python
X, Y =  make_regression(n_samples = 10000, n_features = 10, n_informative = 3)
```

# Splitting the data into training & test

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

```python
len(X_train)
```

```python
len(X_test)
```

# Feature Scaling


```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

# Decision Tree

```python
dt = DecisionTreeRegressor()
dt.fit(X_train,Y_train)
pred = dt.predict(X_test)
print("Decision Tree R2-Score: ", r2_score(pred,Y_test))
print("Decision Tree MEAN Score: ", mean_absolute_error(pred,Y_test))
print("Decision Tree MEAN SQUARE Score: ", mean_squared_error(pred,Y_test))
print("Decision Tree ROOT MEAN SQUARE Score: ", np.sqrt(mean_squared_error(pred,Y_test)))
```

# Bag Classifier

```python
bag = BaggingRegressor(estimator = DecisionTreeRegressor(),
                        n_estimators = 500,
                        max_samples = 0.5,
                        bootstrap = True,
                        random_state = 42)
```

```python
bag.fit(X_train, Y_train)
```

```python
y_pred_bag = bag.predict(X_test)
```

```python
print("Bagging Regressor R2-Score: ", r2_score(y_pred_bag,Y_test))
print("Bagging Regressor MEAN Score: ", mean_absolute_error(y_pred_bag,Y_test))
print("Bagging Regressor MEAN SQUARE Score: ", mean_squared_error(y_pred_bag,Y_test))
print("Bagging Regressor ROOT MEAN SQUARE Score: ", np.sqrt(mean_squared_error(y_pred_bag,Y_test)))
```

# Random Forest Classifier

```python
rf = RandomForestRegressor(n_estimators = 500, random_state = 42) 
rf.fit(X_train,Y_train)
predrf = rf.predict(X_test)
print("Random Forest Regressor R2-Score: ", r2_score(predrf,Y_test))
print("Random Forest Regressor MEAN Score: ", mean_absolute_error(predrf,Y_test))
print("Random Forest Regressor MEAN SQUARE Score: ", mean_squared_error(predrf,Y_test))
print("Random Forest Regressor ROOT MEAN SQUARE Score: ", np.sqrt(mean_squared_error(predrf,Y_test)))
```

# Bagging Using SVM

```python
bagSVM = BaggingRegressor(estimator = SVR(),
                        n_estimators = 500,
                        max_samples = 0.25,
                        bootstrap = True,
                        random_state = 42)
```

```python
bagSVM.fit(X_train, Y_train)
```

```python
y_pred_bagSVM = bagSVM.predict(X_test)
```

```python
print("SVR Bagging Regressor R2-Score: ", r2_score(y_pred_bagSVM,Y_test))
print("SVR Bagging Regressor MEAN Score: ", mean_absolute_error(y_pred_bagSVM,Y_test))
print("SVR Bagging Regressor MEAN SQUARE Score: ", mean_squared_error(y_pred_bagSVM,Y_test))
print("SVR Bagging Regressor ROOT MEAN SQUARE Score: ", np.sqrt(mean_squared_error(y_pred_bagSVM,Y_test)))
```
