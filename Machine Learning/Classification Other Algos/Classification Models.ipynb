```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import classification_report

from sklearn.naive_bayes import BernoulliNB

from sklearn.svm import SVC

from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges, errors = "coerce")      
```

```python
df.info()
```

```python
df.dropna(how = 'any', inplace = True)
```

```python
df.Churn.value_counts()/len(df)*100
```

```python
X = df.drop(columns=["customerID","Churn"])        
```

```python
Y = df.Churn
```

# Feature Encoding

```python
X = pd.get_dummies(X, columns = ['gender','Partner','Dependents','PhoneService','MultipleLines','InternetService','OnlineSecurity',
                                'OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaperlessBilling',
                                'PaymentMethod',], drop_first = True, dtype = 'int')
```

# Splitting the data into training & test


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

# Feature Scaling


```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

# kNN Classifier

```python
knn = KNeighborsClassifier()

knn.fit(X_train, Y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
accuracy_score(y_pred, Y_test)*100
```

# Decision Tree Classifier

```python
tree = DecisionTreeClassifier()
tree.fit(X_train, Y_train)
```

```python
y_pred_tree = tree.predict(X_test)
```

```python
accuracy_score(y_pred_tree, Y_test)*100
```

# Random Forest Classifier

```python
randomForest = RandomForestClassifier()
randomForest.fit(X_train, Y_train)
```

```python
y_pred_random = randomForest.predict(X_test)
```

```python
accuracy_score(y_pred_random, Y_test)*100
```

```python
knn = classification_report(y_pred, Y_test)
dt = classification_report(y_pred_tree, Y_test)
rf = classification_report(y_pred_random, Y_test)
```

```python
print("KNN REPORT: \n", knn)
```

```python
print("DECISION TREE REPORT: \n", dt)
```

```python
print("RANDOM FOREST CLASSIFIER: \n", rf)
```

# Naive Bayes

```python
naiveBayes = BernoulliNB()
naiveBayes.fit(X_train, Y_train)

y_pred_naiveBayes = naiveBayes.predict(X_test)

accuracy_score(y_pred_naiveBayes, Y_test)*100
```

```python
nb = classification_report(y_pred_naiveBayes, Y_test)
print("Naive Bayes: \n", nb)
```

# SVM Classifier

```python
svm = SVC()
svm.fit(X_train, Y_train)

y_pred_svm = svm.predict(X_test)

accuracy_score(y_pred_svm, Y_test)*100
```

```python
svv = classification_report(y_pred_svm, Y_test)
print("SVM: \n", svv)
```

# Logistic Regression

```python
lr = LogisticRegression()
lr.fit(X_train, Y_train)

y_pred_lr = lr.predict(X_test)

accuracy_score(y_pred_lr, Y_test)*100
```

```python
lrr = classification_report(y_pred_lr, Y_test)
print("LOGISTIC REGRESSION: \n", lrr)
```
