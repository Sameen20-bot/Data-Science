```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import classification_report, accuracy_score

from imblearn.over_sampling import RandomOverSampler

from imblearn.under_sampling import RandomUnderSampler

from imblearn.combine import SMOTEENN

from imblearn.over_sampling import SMOTE

from imblearn.over_sampling import ADASYN

from imblearn.under_sampling import AllKNN

from imblearn.under_sampling import TomekLinks
```

```python
df = pd.read_csv(r"Classification Other Algos/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.TotalCharges = pd.to_numeric(df.TotalCharges, errors = "coerce")      
```

```python
df.info()
```

```python
df.dropna(how = 'any', inplace = True)
```

```python
df.Churn.value_counts()/len(df)*100
```

```python
X = df.drop(columns=["customerID","Churn"])        
```

```python
Y = df.Churn
```

# Feature Encoding

```python
X = pd.get_dummies(X, columns = ['gender','Partner','Dependents','PhoneService','MultipleLines','InternetService','OnlineSecurity',
                                'OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaperlessBilling',
                                'PaymentMethod',], drop_first = True, dtype = 'int')
```

# Splitting the data into training & test


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

# Feature Scaling


```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

# Decision Tree Classifier

```python
tree = DecisionTreeClassifier()
tree.fit(X_train, Y_train)
```

```python
y_pred_tree = tree.predict(X_test)
```

```python
accuracy_score(y_pred_tree, Y_test)*100
```

```python
dt = classification_report(y_pred_tree, Y_test)
print("DECISION TREE REPORT: \n", dt)
```

# Imbalance Techniques

## 1. Random Over Sampler

```python
overSampler = RandomOverSampler()
```

```python
X_train_Oversampler, Y_train_Oversampler = overSampler.fit_resample(X_train, Y_train)
```

```python
imbalancedTree = DecisionTreeClassifier()
imbalancedTree.fit(X_train_Oversampler, Y_train_Oversampler)
```

```python
y_pred_tree_embalanced = imbalancedTree.predict(X_test)
```

```python
accuracy_score(y_pred_tree_embalanced, Y_test)*100
```

```python
im_dt = classification_report(y_pred_tree_embalanced, Y_test)
print("IMBALANCED DECISION TREE REPORT: \n", im_dt)
```

## 2. Random Under Sampling

```python
underSampler = RandomUnderSampler()
```

```python
X_train_Undersampler, Y_train_Undersampler = underSampler.fit_resample(X_train, Y_train)
```

```python
imbalancedTree2 = DecisionTreeClassifier()
imbalancedTree2.fit(X_train_Undersampler, Y_train_Undersampler)
```

```python
y_pred_tree_embalanced2 = imbalancedTree2.predict(X_test)
```

```python
im_dt2 = classification_report(y_pred_tree_embalanced2, Y_test)
print("IMBALANCED DECISION TREE REPORT: \n", im_dt2)
```
