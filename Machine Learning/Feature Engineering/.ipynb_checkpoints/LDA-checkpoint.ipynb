```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score

from sklearn.decomposition import PCA

from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
df.head(5)
```

```python
df.info()
```

# Converting TotalCharges to numericals

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df['TotalCharges'].isnull().sum()
```

```python
df.dropna(how='any', inplace=True)
```

```python
df.info()
```

```python
df.drop(['customerID'], axis='columns', inplace=True)
```

```python
df.head()
```

# Dummy Encoding

```python
df = pd.get_dummies(df, drop_first=True, dtype = int)
```

```python
df.head()
```

```python
df.Churn_Yes.value_counts()/len(df)*100
```

```python
X = df.drop(['Churn_Yes'], axis=1)
Y = df['Churn_Yes']
```

# Split the data into training and test sets

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

# Standardizing

```python
sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

# Logistic Model Original

```python
lg = LogisticRegression(max_iter=1000)
lg.fit(X_train_sc, Y_train)
y_pred = lg.predict(X_test_sc)
print("Accuracy of lg model original: ", accuracy_score(Y_test,y_pred)*100)
```

# Applying PCA

```python
pca = PCA(n_components = 10)
X_train_pca = pca.fit_transform(X_train_sc)
X_test_pca = pca.transform(X_test_sc)
```

```python
variance = pca.explained_variance_ratio_
```

```python
variance
```

# Now PCA Based Model

```python
pca_lg = LogisticRegression(max_iter=1000)
pca_lg.fit(X_train_pca, Y_train)
y_pred_pca_lg = pca_lg.predict(X_test_pca)
print("Accuracy of pca_lg model: ", accuracy_score(Y_test,y_pred_pca_lg)*100)
```
