# Lets find the 5 features that are important to Y variable

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score

from sklearn.feature_selection import chi2, SelectKBest

from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
df.head()
```

```python
df.info()
```

# Converting Total Charges to Numeric

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors = "coerce")
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df.dropna(inplace = True)
```

```python
df.info()
```

```python
df = df.drop(columns = ["customerID"])
```

# Dummy Encoding

```python
df = pd.get_dummies(df, drop_first = True, dtype = int)
```

```python
df.head()
```

```python
df.Churn_Yes.value_counts()/len(df)*100
```

```python
X = df.drop(['Churn_Yes'], axis=1)
y = df['Churn_Yes']
```

# Split the data into training and test sets


```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Standardizing

```python
sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

# Logistic Model Original

```python
lg = LogisticRegression(max_iter=1000)
lg.fit(X_train_sc, y_train)
y_pred = lg.predict(X_test_sc)
print("Accuracy of lg model original: ", accuracy_score(y_test,y_pred)*100)
```

# Applying CHI2

```python
mm = MinMaxScaler()
X_train_mm = mm.fit_transform(X_train)
X_test_mm = mm.transform(X_test)
```

```python
chi2_selector = SelectKBest(chi2, k = 5)
X_train_chi2 = chi2_selector.fit_transform(X_train_mm, y_train)
X_test_sc_chi2 = chi2_selector.transform(X_test_mm)
```

# Build Chi2 model

```python
chi_lg = LogisticRegression(max_iter=1000)
chi_lg.fit(X_train_chi2, y_train)
chi_y_pred = chi_lg.predict(X_test_sc_chi2)
print("Accuracy of lg model original: ", accuracy_score(y_test,chi_y_pred)*100)
```
