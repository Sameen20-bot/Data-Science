```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.datasets import make_classification

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import accuracy_score

from sklearn.ensemble import BaggingClassifier, RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier

from sklearn.tree import DecisionTreeClassifier

from sklearn.svm import SVC

from xgboost import XGBClassifier
```

```python
### n_informative make_classification() mein batata hai ke total features mein se kitne 
### features actually target (y) ko predict karne ke liye useful/informative hain.
```

```python
X, Y =  make_classification(n_samples = 10000, n_features = 10, n_informative = 3)
```

# Splitting the data into training & test

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

```python
len(X_train)
```

```python
len(X_test)
```

# Feature Scaling


```python
sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.transform(X_test)
```

```python
X_train_sc
```

```python
X_test_sc
```

# Decision Tree

```python
dt = DecisionTreeClassifier()
dt.fit(X_train_sc,Y_train)
pred = dt.predict(X_test_sc)
print(accuracy_score(pred,Y_test)*100)
```

# Bag Classifier

```python
bag = BaggingClassifier(estimator = DecisionTreeClassifier(),
                        n_estimators = 500,
                        max_samples = 0.5,
                        bootstrap = True,
                        random_state = 42)
```

```python
bag.fit(X_train_sc, Y_train)
```

```python
y_pred_bag = bag.predict(X_test_sc)
```

```python
print(accuracy_score(y_pred_bag,Y_test)*100)
```

# Random Forest Classifier

```python
rf = RandomForestClassifier(n_estimators = 500, random_state = 42) 
rf.fit(X_train_sc,Y_train)
predrf = rf.predict(X_test_sc)
print(accuracy_score(predrf,Y_test)*100)
```

# Bagging Using SVM

```python
bagSVM = BaggingClassifier(estimator = SVC(),
                        n_estimators = 500,
                        max_samples = 0.25,
                        bootstrap = True,
                        random_state = 42)
```

```python
bagSVM.fit(X_train_sc, Y_train)
```

```python
y_pred_bagSVM = bagSVM.predict(X_test_sc)
```

```python
print(accuracy_score(y_pred_bagSVM,Y_test)*100)
```

# Boosting

## 1. Ada Boosting

```python
ada = AdaBoostClassifier(n_estimators = 500)
ada.fit(X_train_sc,Y_train)
predada = ada.predict(X_test_sc)
print(accuracy_score(predada,Y_test)*100)
```

## 2. Gradient Boosting

```python
gb = GradientBoostingClassifier(n_estimators = 500)
gb.fit(X_train_sc,Y_train)
predgb = gb.predict(X_test_sc)
print(accuracy_score(predgb,Y_test)*100)
```

## Blindly go for Boosting Algorithms bcuz they are best, especially XGBoost (Abhi HPO krega phir or acha malom hoga)

## 2. XG Boosting

```python
xb = XGBClassifier(n_estimators = 500)
xb.fit(X_train_sc,Y_train)
predxb = xb.predict(X_test_sc)
print(accuracy_score(predxb,Y_test)*100)
```
