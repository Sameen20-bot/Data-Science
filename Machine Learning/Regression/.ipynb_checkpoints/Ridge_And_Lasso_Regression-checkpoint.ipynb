```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression, Ridge, Lasso

from sklearn.metrics import r2_score

from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("boston.csv")
```

```python
df.head()
```

```python
X = df.drop(columns=["Price"])
```

```python
Y = df.Price.values
```

```python
X
```

```python
Y
```

# Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
lasso = Lasso(alpha = 1.0)
lasso.fit(X_train, Y_train)
```

```python
bad_columns = X_train.columns[np.where(lasso.coef_ == 0)[0]]
bad_columns
```

```python
X_train_clean = X_train.drop(columns = bad_columns)
X_test_clean = X_test.drop(columns = bad_columns)
```

# Standardization

```python
sc = StandardScaler()
```

```python
X_train_clean = sc.fit_transform(X_train_clean)
X_test_clean = sc.transform(X_test_clean)
```

```python
X_train_clean
```

```python
X_test_clean
```

# Linear Regression Model

```python
lr = LinearRegression()
lr.fit(X_train_clean, Y_train)
```

# Calculate R2 Score

```python
y_pred_lr = lr.predict(X_test_clean)
```

```python
r2_score(Y_test, y_pred_lr)
```

# Ridge And Lasso

### Applying Ridge

```python
ridge_model = Ridge(alpha = 1.0)
ridge_model.fit(X_train, Y_train)

y_pred_ridge = ridge_model.predict(X_test)
r2_score(Y_test, y_pred_ridge)
```

```python
ridge_model1 = Ridge(alpha = 0.5)
ridge_model1.fit(X_train, Y_train)

y_pred_ridge1 = ridge_model1.predict(X_test)
r2_score(Y_test, y_pred_ridge1)
```

```python
ridge_model2 = Ridge(alpha = 0.2)
ridge_model2.fit(X_train, Y_train)

y_pred_ridge2 = ridge_model2.predict(X_test)
r2_score(Y_test, y_pred_ridge2)
```

### Applying Lasso

```python
lasso = Lasso(alpha = 1.0)
lasso.fit(X_train, Y_train)

y_pred_lasso = lasso.predict(X_test)
r2_score(Y_test, y_pred_lasso)
```
