```python
import pandas as pd
import numpy as np
from sklearn import preprocessing
```

```python
df = pd.read_csv("Churn_Modelling.csv")
df.head()
```

```python
df.info()
```

## Data Cleaning

```python
df["Gender"] = df["Gender"].fillna("Male")
```

```python
df["Age"] = df["Age"].fillna(df.Age.mean())
```

```python
df.info()
```

```python
df.drop(columns=["RowNumber","CustomerId","Surname"],inplace=True,axis=1)               
```

```python
df.head()
```

# Label Encoding

```python
le = preprocessing.LabelEncoder()
df["Gender_Label"] = le.fit_transform(df.Gender.values)
```

```python
df.head()
```

```python
df.drop(columns=["Gender"],inplace=True,axis=1)
```

# One Hot Encoding

```python
one_hot = pd.get_dummies(df, dtype=int)
one_hot.head()
```

# Dummy Encoding

```python
dummy = pd.get_dummies(df, dtype=int, drop_first = True)
dummy.head()
```
